```python

from google.colab import drive
drive.mount('/content/drive')

import os
import json
import urllib.request
import urllib.parse

class FlowerDownloader:

    def __init__(self):
        self.folder = '/content/drive/MyDrive/project/raw_images'
        self.flowers = {
            'roses': 'rose flower',
            'sunflowers': 'sunflower',
            'tulips': 'tulip flower',
            'daisies': 'daisy flower'
        }

    def search_images(self, keyword):
        params = {
            'action': 'query',
            'generator': 'search',
            'gsrsearch': 'filetype:bitmap ' + keyword,
            'gsrnamespace': 6,
            'gsrlimit': 50,
            'prop': 'imageinfo',
            'iiprop': 'url',
            'iiurlwidth': 500,
            'format': 'json'
        }

        url = 'https://commons.wikimedia.org/w/api.php?'
        url += urllib.parse.urlencode(params)

        request = urllib.request.Request(
            url,
            headers={'User-Agent': 'FlowerProject/1.0 (educational use)'}
        )

        with urllib.request.urlopen(request, timeout=30) as response:
            data = json.loads(response.read().decode('utf-8'))

        urls = []

        for page in data.get('query', {}).get('pages', {}).values():
            info = page.get('imageinfo', [])

            if info:
                urls.append(info[0].get('thumburl', info[0].get('url')))

        return urls

    def download(self, url, folder, number):
        try:
            request = urllib.request.Request(
                url,
                headers={'User-Agent': 'FlowerProject/1.0'}
            )

            with urllib.request.urlopen(request, timeout=30) as response:
                image_data = response.read()
                content_type = response.headers.get('Content-Type', '')

            if not content_type.startswith('image/'):
                return False

            if len(image_data) < 1000:
                return False

            extension = '.jpg'
            if 'png' in content_type:
                extension = '.png'
            elif 'webp' in content_type:
                extension = '.webp'
            elif 'gif' in content_type:
                extension = '.gif'

            filename = 'flower_' + str(number) + extension
            path = os.path.join(folder, filename)

            if os.path.exists(path):
                return False

            with open(path, 'wb') as file:
                file.write(image_data)

            return True

        except Exception:
            return False

    def collect(self):
        os.makedirs(self.folder, exist_ok=True)
        total = 0

        for category, keyword in self.flowers.items():
            category_folder = os.path.join(self.folder, category)
            os.makedirs(category_folder, exist_ok=True)

            print('Collecting:', category)

            try:
                urls = self.search_images(keyword)
            except Exception as error:
                print('Search failed:', error)
                continue

            count = 0

            for url in urls:
                if count >= 25:
                    break

                if self.download(url, category_folder, count + 1):
                    count += 1
                    total += 1
                    print('Saved', category, count)

            print(category, 'images saved:', count)

        print('Total new images downloaded:', total)
        print('Drive folder:', self.folder)

downloader = FlowerDownloader()
downloader.collect()
```

```python

import os
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

class FlowerProject:

    def __init__(self):
        self.folder = '/content/drive/MyDrive/project/raw_images'
        self.categories = ['roses', 'sunflowers', 'tulips', 'daisies']
        self.images = []
        self.labels = []

    def load_images(self):
        for category in self.categories:
            category_folder = os.path.join(self.folder, category)
            count = 0

            if not os.path.exists(category_folder):
                print("Folder not found:", category_folder)
                continue

            for filename in os.listdir(category_folder):
                path = os.path.join(category_folder, filename)

                try:
                    image = Image.open(path).convert('RGB')
                    image = image.resize((128, 128))

                    self.images.append(np.array(image))
                    self.labels.append(category)
                    count += 1

                except Exception:
                    print("Could not load:", filename)

            print(category, ":", count, "images loaded")

        print("Total images loaded:", len(self.images))

    def preprocess_images(self):
        self.images = np.array(self.images)

        if len(self.images) > 0:
            self.images = self.images / 255.0

        print("Preprocessing complete.")

        if len(self.images) > 0:
            print("Array shape:", self.images.shape)

    def visualize_images(self):
        if len(self.images) == 0:
            print("No images available to display.")
            return

        number = min(8, len(self.images))

        plt.figure(figsize=(12, 6))

        for i in range(number):
            plt.subplot(2, 4, i + 1)

            image = self.images[i]

            plt.imshow(image)
            plt.title(self.labels[i])
            plt.axis('off')

        plt.tight_layout()
        plt.show()


flower_project = FlowerProject()

flower_project.load_images()
flower_project.preprocess_images()
flower_project.visualize_images()
```
